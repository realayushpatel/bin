```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('datafile (1).csv',sep=",")
```

```python
sns.countplot(x='2010-2011',data=df,palette='bwr')
plt.xticks(rotation=45)
```

```python
sns.countplot(x='2009-10',data=df,palette='bwr')
plt.xticks(rotation=45)
```

```python
df['Total'] = df[['2010-2011', '2009-10', '2008-09']].sum(axis=1)
df_sorted = df.sort_values('Total', ascending=False)
sns.barplot(data=df_sorted, x='States/UTs', y='Total', palette='viridis')
plt.xticks(rotation=90)
```

```python
sns.lineplot(data=df, x='States/UTs', y='2010-2011')
plt.xticks(rotation=90)
```

```python
sns.lineplot(data=df, x='States/UTs', y='2008-09')
plt.xticks(rotation=90)
```

```python
df_num=df.select_dtypes(include='number')
sns.heatmap(df_num.corr(),annot=True,cmap='YlGnBu')
```

```python
sns.clustermap(df_num.corr(),cmap='coolwarm')
```

```python
sns.histplot(data=df, x='2010-2011', bins=10, kde=True)
```
